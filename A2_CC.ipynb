```python
import random
from collections import Counter

# Q1: Lists
roll_number = "1024160136"
digits = [int(d) for d in roll_number]
L = [d * 10 for d in digits]
L_original = L.copy()

# i. Print L
print("i.  L =", L)

# ii. Add two numbers - one with append(), one with insert()
L.append(99)
print("ii. After append(99):", L)

L.insert(2, 55)
print("ii. After insert(2, 55):", L)

# iii. Remove two elements - one with remove(), one with pop()
L.remove(0)
print("iii. After remove(0):", L)

L.pop(3)
print("iii. After pop(3):", L)

# iv. Sort ascending, then descending
L.sort()
print("iv. Ascending:", L)
L.sort(reverse=True)
print("iv. Descending:", L)

# v. Slicing - first three and last three elements
print("v. First three:", L[:3])
print("v. Last three:", L[-3:])

# vi. List comprehension - elements greater than the average
avg_L = sum(L) / len(L)
above_avg = [x for x in L if x > avg_L]
print("vi. Average =", avg_L, "| Elements above average:", above_avg)
```

```python
# Q2: Tuples
scores = tuple(L_original[:8])
print("\nScores tuple:", scores)

# i. Highest score + its index, lowest score + its frequency
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print("i.  Highest:", highest, "at index", highest_index)
print("i.  Lowest:", lowest, "appears", lowest_count, "times")

# ii. Reverse the tuple, return as a list
reversed_scores = list(reversed(scores))
print("ii. Reversed (as list):", reversed_scores)

# iii. Ask user for a score, find first occurrence index
user_score = int(input("iii. Enter a score to search for: "))
if user_score in scores:
    print("iii. Found at index:", scores.index(user_score))
else:
    print("iii. Score not present in tuple.")

# iv. Attempt to modify a tuple element -> capture the error
try:
    scores[0] = 100
except TypeError as e:
    print("iv. Error caught:", e)

# v. Unpack into first, second, and the rest
first_score, second_score, *remaining_scores = scores
print("v. First:", first_score, "| Second:", second_score, "| Remaining:", remaining_scores)
```

```python
# Q3: Random numbers with seed
# i. A list of 100 random numbers between 100 and 900 (inclusive)
random.seed(1024160136)
random_list = [random.randint(100, 900) for _ in range(100)]

# ii. Odd numbers
odd_count = len([n for n in random_list if n % 2 != 0])
print("\nii. Odd numbers count:", odd_count)

# iii. Even numbers
even_count = len([n for n in random_list if n % 2 == 0])
print("iii. Even numbers count:", even_count)

# iv. Prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_list = [n for n in random_list if is_prime(n)]
print("iv. Prime numbers count:", len(prime_list))
print("iv. Prime numbers list:", prime_list)

# v. Most frequent number
freq = Counter(random_list)
most_common_num, most_common_count = freq.most_common(1)[0]
print("v. Most frequent number:", most_common_num, "| Occurs:", most_common_count, "times")
```

```python
# Q4: Sets
digits8 = digits[:8]
A = {d * 7 for d in digits8}
B = {d * 9 for d in digits8}
print("\nSet A:", A)
print("Set B:", B)

# vi. Union
print("vi. Union A ∪ B:", A | B)

# vii. Intersection
print("vii. Intersection A ∩ B:", A & B)

# viii. Difference A-B and B-A
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)
print("viii. A - B:", diff_A_B)
print("viii. B - A:", diff_B_A)

# ix. Symmetric difference
print("ix. Symmetric difference:", A.symmetric_difference(B))

# x. Subset / superset checks
print("x. Is A subset of B?", A.issubset(B))
print("x. Is B superset of A?", B.issuperset(A))

# xi. Discard user input X from A
X = int(input("xi. Enter a value to discard from set A: "))
A.discard(X)
print("xi. Set A after discard:", A)
```

```python
# Q5: Dictionaries
my_dict = {
    "name": "Pridhi Thareja",
    "roll_no": "1024160136",
    "branch": "Computer Science and Engineering",
    "age": 19,
    "city": "Patiala"
}

# i. Rename "city" to "location" (works for any dictionary)
my_dict["location"] = my_dict.pop("city")
print("\ni.  After renaming city->location:", my_dict)

# ii. Add "cgpa"
my_dict["cgpa"] = 8
print("ii. After adding cgpa:", my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("iii. After incrementing age:", my_dict)

# iv. Delete "branch" using pop() and del, in two separate copies
dict_pop_copy = my_dict.copy()
popped_value = dict_pop_copy.pop("branch")
print("iv. pop() removed branch, returned value:", popped_value, "| Dict now:", dict_pop_copy)
dict_del_copy = my_dict.copy()
del dict_del_copy["branch"]
print("iv. del removed branch (no return value) | Dict now:", dict_del_copy)

# v. Iterate with .items()
print("v. Key -> Value pairs:")
for key, value in my_dict.items():
    print(f"   {key} \u2192 {value}")

# vi. Safely check for "email" key
if "email" in my_dict:
    print("vi. Email:", my_dict["email"])
else:
    print("vi. No email found for this user.")

# vii. Merge with a friend's dictionary
friend_dict = {
    "name": "Aanya Kapoor",
    "roll_no": "1024160199",
    "branch": "Electronics and Communication Engineering",
    "age": 20,
    "city": "Chandigarh"
}
merged_dict = {**my_dict, **friend_dict}
print("vii. Merged dict:", merged_dict)

# viii. Dictionary comprehension - only string values
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("viii. String-valued pairs only:", string_only_dict)
```
