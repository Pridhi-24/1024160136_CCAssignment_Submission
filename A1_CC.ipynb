```python
# Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.
n = int(input("Enter a number (n): "))
total_sum = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        total_sum += i
print(f"The sum of numbers divisible by both 7 and 9 from 1 to {n} is: {total_sum}")
```

```python
# Assignment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.
n = int(input("Enter a number (n): "))
prime_sum = 0
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num
print(f"The sum of all prime numbers from 1 to {n} is: {prime_sum}")
```

```python
# Assignment 6.1: WAP using function that add all odd numbers from 1 to n
def sum_odd_numbers(n):
    total_sum = 0
    for i in range(1, n + 1, 2):
        total_sum += i
    return total_sum
n = int(input("Enter a number (n): "))
result = sum_odd_numbers(n)
print(f"The sum of all odd numbers from 1 to {n} is: {result}")
```

```python
# Assignment 6.2: WAP using function that add all prime numbers from 1 to n
def is_prime(num):
    """Helper function to check if a single number is prime."""
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
def sum_prime_numbers(n):
    """Function to sum all prime numbers up to n."""
    prime_sum = 0
    for i in range(2, n + 1):
        if is_prime(i):
            prime_sum += i
    return prime_sum
n = int(input("Enter a number (n): "))
result = sum_prime_numbers(n)
print(f"The sum of all prime numbers from 1 to {n} is: {result}")
```
